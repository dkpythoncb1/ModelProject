# Part 1: A consumer with two nests

Before solving the model (from part 2 onward), this section restates the setup from the
assignment and explains the reparametrization the whole project relies on.

**The problem.** The consumer buys food ($x_1$), bus trips ($x_2$) and train trips ($x_3$).
Bus and train sit together in one nest ("travel"), because they are much closer substitutes
for each other than either one is for food. The nested CES aggregator first combines bus and
train into a travel composite $x_B$ (eq. 1 in the assignment), then combines food and travel
into utility (eq. 2). Both nests use the same functional form — `.ces()` in `Consumer.py` —
just with different weights ($\alpha,\beta$) and substitution parameters ($\sigma_A,\sigma_B$).

**Nested budget shares.** Rather than choosing $(x_1,x_2,x_3)$ directly under the constraint
$p_1x_1+p_2x_2+p_3x_3=I$, we choose two numbers in $[0,1]$: $s_1$, the share of income spent
on food, and $w$, the share of the *remaining* travel budget spent on the bus. The three
budget shares then follow mechanically as $s_1$, $(1-s_1)w$ and $(1-s_1)(1-w)$, which always
sum to one — so the budget constraint holds automatically for *any* $(s_1,w)\in[0,1]\times[0,1]$.
That is why the whole feasible set is the unit square, and why `solve()` can hand `L-BFGS-B`
the box `bounds=((0,1),(0,1))`, with no penalty function or other constraint handling anywhere
in this project.

**The two calibrations.** We use $\sigma_B=0.40$ (bus and train are complements: the consumer
wants both, and cannot easily replace one with the other) and $\sigma_B=3.00$ (bus and train
are substitutes: either one will do), holding every other parameter fixed at the values in the
table above.

**How we know an answer is right (section 1.3).** Two cheap checks run through the whole
project: (1) *is the answer possible* — all three shares strictly between 0 and 1 and summing
to one, all three quantities positive; (2) *do two different methods agree* — grid search and
L-BFGS-B should land on (near) the same point. We check both explicitly once the model is
solved: check 1 right after the grid search in section 2.1, and check 2 in section 2.2, where
the two methods are compared directly.

# Part 2: Solving the model numerically

We solve the consumer problem from part 1 in two ways — a two-dimensional grid search and
`scipy.optimize.minimize` with `L-BFGS-B` — and compare them.

## 2.1 Two-dimensional grid search

We lay a grid over everything the consumer can afford (the unit square $[0,1]\times[0,1]$ in
$(s_1,w)$), compute utility in every point, and keep the best one. There is nothing to mask
out, and no `np.nan`: every point of the square is a possible choice. This is implemented as
`solve_grid()` in Consumer.py.

```python
# solve by grid search for both calibrations, and store the results
# (opt.s1, opt.w, opt.s2, opt.s3, opt.u, plus the grids for plotting) in a dict
grid_results = {}
for name, model in models.items():
    print(name)
    # N=200 grid points in each dimension -- a reasonable baseline resolution
    grid_results[name] = model.solve_grid(N=200, do_print=True)
    print()
```

Both calibrations give a possible solution: all three budget shares lie strictly
between 0 and 1 and sum to one (check 1 from section 1.3).

### Figure: utility over the nested shares, in 3D and as a contour plot

```python
# plot utility over (s1,w) for both calibrations: a 3D surface (left column)
# and the same data as a contour plot (right column), with the solution marked
fig = plt.figure(figsize=(12, 10))

for i, (name, model) in enumerate(models.items()):
    opt = grid_results[name]   # the grid-search result for this calibration

    # --- left column: 3D surface plot ---
    ax = fig.add_subplot(2, 2, 2 * i + 1, projection='3d')
    ax.plot_surface(opt.s1_grid, opt.w_grid, opt.u_grid, cmap='viridis',
                     linewidth=0, antialiased=True, alpha=0.9, zorder=1)
    # mark the grid-search solution with a red dot, drawn on top of the surface
    ax.scatter(opt.s1, opt.w, opt.u, color='red', s=60, zorder=10)
    ax.set_xlabel('$s_1$'); ax.set_ylabel('$w$'); ax.set_zlabel('$u$')
    ax.set_xlim(0, 1); ax.set_ylim(0, 1)   # so the two axes meet at (0,0), as in a 2D plot
    ax.set_title(f'{name}\n3D surface')

    # --- right column: contour plot of the same utility surface ---
    ax2 = fig.add_subplot(2, 2, 2 * i + 2)
    cs = ax2.contourf(opt.s1_grid, opt.w_grid, opt.u_grid, levels=30, cmap='viridis')
    ax2.plot(opt.s1, opt.w, 'ro', markersize=9, label='solution', zorder=10)
    ax2.set_xlabel('$s_1$'); ax2.set_ylabel('$w$')
    ax2.set_xlim(0, 1); ax2.set_ylim(0, 1)
    ax2.set_title(f'{name}\nContour plot')
    ax2.legend(loc='upper right')
    fig.colorbar(cs, ax=ax2, label='$u$')

plt.tight_layout()
fig.savefig('fig_2_1_grid.png', dpi=110)   # save the figure to disk
plt.show()
```

**Reading the contour plot.** Where the contour lines sit close together, utility changes a
lot over a small move in $(s_1,w)$ — the surface is steep there; where they are spread out, it
is flat. In both panels the lines fan out from a single peak near the middle of the square
(the solution) and pack tightly close to the edges $w=0$ and $w=1$, i.e. spending almost
nothing on one of the two travel goods. That follows directly from the nested CES form: with
$\sigma_B<1$ (complements, left panel), $\rho_B=1-1/\sigma_B<0$, so the travel composite
$x_B=\big(\beta x_2^{\rho_B}+(1-\beta)x_3^{\rho_B}\big)^{1/\rho_B}$ blows up as either $x_2$ or
$x_3$ approaches zero, and utility punishes that hard — hence the tightly packed contours near
the corners. With $\sigma_B>1$ (substitutes, right panel) the same corner costs much less,
because the consumer can lean almost entirely on the cheaper travel good without losing much
utility — hence the visibly wider spacing near $w=0$ and $w=1$ there.

### How much does the answer move as we add grid points?

We try $N=50,100,500,1000$ and look at how much the solution moves, and how many
evaluations of $u$ it costs ($N^2$).

```python
# for each calibration, solve on grids of increasing resolution and record:
# the solution, how far it moved from the previous (coarser) grid, and the time taken
rows = []
for name, model in models.items():
    previous = None   # the (s1, w) from the previous, coarser grid
    for N in [50, 100, 500, 1000]:
        t0 = time.time()
        opt = model.solve_grid(N=N, do_print=False)
        dt = time.time() - t0

        # euclidean distance between this solution and the previous grid's solution
        if previous is None:
            move = np.nan
        else:
            move = np.sqrt((opt.s1 - previous[0])**2 + (opt.w - previous[1])**2)

        rows.append({
            'Calibration': name, 'N': N, 'Evaluations': N * N,
            's1': opt.s1, 'w': opt.w, 'u': opt.u,
            'Move from previous N': move, 'Time (s)': dt,
        })
        previous = (opt.s1, opt.w)

# collect everything in a table for easy reading
df_N = pd.DataFrame(rows)
df_N
```

**Answer:** The solution does not move monotonically or smoothly with $N$ -- it jumps
around with small, shrinking steps, because the grid solution always lands on the grid point
closest to the true optimum. From $N=50$ to $N=1000$, $(s_1,w)$ changes by less than $0.01$ in
both calibrations, while the number of evaluations grows from $2{,}500$ to $1{,}000{,}000$ (a
factor of $400$). The utility level $u$ converges quickly and smoothly (it is simply the
*best of many points*, and it can only get better, never worse, as $N$ increases), but
$(s_1,w)$ itself converges *unevenly*, because the grid points don't sit exactly on the
optimum. In other words, there is a clear diminishing-returns pattern: a computation that is
400 times more expensive at $N=1000$ only buys a handful of extra decimal places of precision
compared with $N=50$.

### Comparing the two calibrations: where is the surface steep, and where is it flat?

**Answer:** As the contour plot above shows, both surfaces are flattest right around the
solution (which is exactly what makes it an optimum) and steepest near the edges of the
square, especially close to $w=0$ or $w=1$. The complements calibration ($\sigma_B=0.40$) is
visibly steeper there than the substitutes calibration ($\sigma_B=3.00$) — the contour lines
pack together much more tightly near the corners in the left-hand panel. The mechanism is the
one described above: with complements, a near-zero quantity of either travel good sends the
CES aggregator toward its worst possible value, so utility drops sharply; with substitutes,
the consumer loses comparatively little from relying on just one of the two.

A broad, flat region around the optimum (substitutes) is, in principle, *easier* for a
gradient-based optimizer: the gradient points reliably toward the optimum over a wide area. A
narrow, steep region (complements) can instead create sharp ridges near the boundary where the
gradient changes direction quickly — and that is exactly what happens in section 2.2.3 below,
where two of the six starting points fail in the complements calibration but none fail in the
substitutes calibration.

## 2.2 L-BFGS-B

L-BFGS-B is the workhorse for smooth problems with a few variables and simple bounds. Because
we wrote the problem in nested budget shares, the bounds ((0,1),(0,1)) are the entire
constraint -- there is no penalty function or any other constraint handling anywhere. This is
implemented as `solve()` in Consumer.py.

```python
# solve with L-BFGS-B for both calibrations, and store the results
# (opt.s1, opt.w, opt.s2, opt.s3, opt.u, opt.path, opt.res) in a dict
lbfgsb_results = {}
for name, model in models.items():
    print(name)
    lbfgsb_results[name] = model.solve(do_print=True)
    print()
```

### Comparison with the grid search

```python
# compare the grid search (at a fairly fine N=500) against L-BFGS-B:
# solution, number of function evaluations, and running time
rows = []
for name, model in models.items():
    # time the grid search
    t0 = time.time(); opt_g = model.solve_grid(N=500, do_print=False); tg = time.time() - t0
    # time L-BFGS-B
    t0 = time.time(); opt_s = model.solve(do_print=False);            ts = time.time() - t0

    rows.append({
        'Calibration': name, 'Method': 'Grid search (N=500)',
        's1': opt_g.s1, 'w': opt_g.w, 'u': opt_g.u,
        'Evaluations': 500 * 500, 'Time (s)': tg,
    })
    rows.append({
        'Calibration': name, 'Method': 'L-BFGS-B',
        's1': opt_s.s1, 'w': opt_s.w, 'u': opt_s.u,
        'Evaluations': opt_s.res.nfev, 'Time (s)': ts,
    })

df_compare = pd.DataFrame(rows)
df_compare
```

**Answer:** The two methods agree on the solution to 3-4 decimal places in both
calibrations -- this is exactly the second check from section 1.3 ("do two different methods
agree?"), and they pass it. But L-BFGS-B gets there using roughly $10{,}000$ times fewer
evaluations of $u$ (18-27 vs. 250,000) and is much faster. The grid search is robust and easy
to understand, but it scales very poorly: it costs $N^2$ evaluations for only $N$ points in
each direction, while L-BFGS-B uses gradient information to head straight for the top.

### The convergence path

We record the convergence path using the `callback` argument of `optimize.minimize`, and draw
it on top of the contour plot from section 2.1. Note that an *iteration* is not the same as a
*function evaluation*: the solver computes $u$ several times per iteration (for the gradient
and the line search), and the callback only fires for the steps it actually takes. The starting
point is not included in the callback, so we add `s0` to the list ourselves before calling the
solver.

```python
# plot the convergence path (from Consumer.py's solve()) on top of the
# contour plot from section 2.1, for both calibrations
fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))

for ax, (name, model) in zip(axes, models.items()):
    opt_g = grid_results[name]     # grid results, used just for the contour background
    opt_s = lbfgsb_results[name]   # L-BFGS-B results, including the recorded path
    path = opt_s.path              # array of shape (n_iterations+1, 2)

    # background: the utility surface as a contour plot
    cs = ax.contourf(opt_g.s1_grid, opt_g.w_grid, opt_g.u_grid, levels=30, cmap='viridis')
    # the path the solver actually took, one point per iteration, drawn on top of it
    ax.plot(path[:, 0], path[:, 1], 'o-', color='white', markersize=5,
            linewidth=1.5, label='convergence path', zorder=5)
    ax.plot(path[0, 0], path[0, 1], 's', color='red', markersize=9, label='start', zorder=10)
    ax.plot(opt_s.s1, opt_s.w, '*', color='orange', markersize=16, label='solution', zorder=10)
    ax.set_xlabel('$s_1$'); ax.set_ylabel('$w$')
    ax.set_xlim(0, 1); ax.set_ylim(0, 1)
    ax.set_title(name)
    ax.legend(loc='upper right', fontsize=8)

plt.tight_layout()
fig.savefig('fig_2_2_path.png', dpi=110)
plt.show()
```

```python
# plot how far each iteration is from the point the solver ends at,
# against the iteration number, on a log scale
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

for ax, (name, model) in zip(axes, models.items()):
    opt_s = lbfgsb_results[name]
    final_point = opt_s.path[-1]                                    # where the solver stopped
    diff = opt_s.path - final_point
    distance = np.sqrt(diff[:, 0]**2 + diff[:, 1]**2)    # distance to that point, per iteration

    # the last point has distance 0, which is undefined on a log scale -- drop it
    ax.plot(range(len(distance) - 1), distance[:-1], 'o-')
    ax.set_yscale('log')
    ax.set_xlabel('iteration')
    ax.set_ylabel('distance to final point (log scale)')
    ax.set_title(name)
    ax.grid(True, which='both', alpha=0.3)

plt.tight_layout()
fig.savefig('fig_2_2_convergence.png', dpi=110)
plt.show()
```

**Answer:** Reading the plot carefully — the y-axis is on a log scale, and the x-axis counts
iterations, not function evaluations — the picture is closer to the opposite of "fast at the
start, slow at the end". In the complements calibration the distance to the final point falls
by only about 30% on the very first step, and by roughly another 30% on the second — genuinely
slow, close-to-linear progress. Only on the last two steps does it fall by an order of
magnitude or more per iteration. The substitutes calibration shows the same shape over its
extra two iterations: three moderate steps (each cutting the distance by only 25-35%), then a
collapse of more than an order of magnitude per iteration over the final three.

That pattern — slow, roughly linear progress far from the optimum, then a sudden acceleration
close to it — is exactly what the theory behind L-BFGS-B predicts, not its reverse.
Superlinear convergence is a *local* property: the algorithm builds up an approximation of the
curvature (the Hessian) from the sequence of gradients it has seen, and that approximation
only becomes accurate once the iterates are already close to the solution. Far from the
optimum it behaves more like a slower, first-order method; near it, the improving curvature
estimate lets it close in almost immediately. This also explains why the complements
calibration needs fewer iterations overall (4, against 6 for substitutes): its steeper surface
(section 2.1) gives the solver a stronger, more informative gradient signal throughout, so it
reaches the region where fast local convergence kicks in sooner.

### Six different starting points

We solve from the middle $(\tfrac12,\tfrac12)$ and all four corners of the square, plus one
extra point close to an edge, for both calibrations.

```python
# six starting points: the middle, all four corners, and one point near an edge
starting_points = {
    'middle (0.5, 0.5)': (0.5, 0.5),
    'corner (0, 0)':     (0.0, 0.0),
    'corner (0, 1)':     (0.0, 1.0),
    'corner (1, 0)':     (1.0, 0.0),
    'corner (1, 1)':     (1.0, 1.0),
    'edge (0.1, 0.9)':   (0.1, 0.9),
}

rows = []
for name, model in models.items():
    for point_name, s0 in starting_points.items():
        # solve with L-BFGS-B from this particular starting point
        opt = model.solve(s0=np.array(s0), do_print=False)
        rows.append({
            'Calibration': name, 'Starting point': point_name,
            's1': opt.s1, 'w': opt.w, 'u': opt.u,
            'Iterations': opt.res.nit, 'Evaluations': opt.res.nfev,
            'success': opt.res.success, 'Message': opt.res.message,
        })

df_start = pd.DataFrame(rows)
df_start
```

**Answer:** In the substitutes calibration, all six starting points find the same
solution ($u\approx 3.4851$), though with a varying number of iterations (most from the
corners, fewest from the middle and the edge point). In the complements calibration, four out
of six starting points find the correct solution ($u\approx 3.4017$) -- but **the corners
$(1,0)$ and $(1,1)$ fail**: the solver stops after just 1 iteration with $u=0$, and still
reports `success = True`.

**What happens at that corner?** At $s_1=1$, the entire income goes to food, and the travel
budget is zero regardless of $w$. The quantities of bus and train become so small that they hit
the floor `par.s_min = 1e-12` in `.ces()`. Because $\sigma_B=0.40<1$ in the complements
calibration, $\rho_A<0$, and a near-zero travel composite produces a *huge* CES term, which --
once raised to the power $1/\rho_A<0$ -- gets pushed all the way down toward $u\approx 0$
instead of up. The numerical gradient that L-BFGS-B estimates with a small step
(`eps`$\approx 10^{-8}$) is, in practice, flat around this corner, because both neighbours of
$s_1=1$ hit the same floor value or a similarly extreme level. The solver interprets that flat
gradient as convergence after a single step and stops -- reporting `success = True` -- at a
point that is actually far from the true optimum ($u\approx3.40$ vs. $u=0$). This is exactly
the kind of error that check 1 in section 1.3 ("is the answer possible?") would catch: here
$s_2=s_3=0$, not strictly positive, so the answer fails that test -- which is also why, in
practice, we should always compare several starting points and both solution methods, rather
than trusting a single `success = True` blindly.

### The effect of `ftol` and `gtol`

We vary the two stopping criteria and report the shares, the utility reached, and the number
of function evaluations for each setting, starting from the middle $(\tfrac12,\tfrac12)$.

```python
# vary ftol (stop when the objective stops changing) and gtol (stop when the
# gradient is small), and record the solution and cost for each setting
rows = []
for name, model in models.items():
    for ftol in [1e-4, 1e-8, 1e-12, 1e-16]:
        opt = model.solve(do_print=False, options={'ftol': ftol})
        rows.append({'Calibration': name, 'Setting': 'ftol', 'Value': ftol,
                      's1': opt.s1, 'w': opt.w, 'u': opt.u,
                      'Evaluations': opt.res.nfev, 'success': opt.res.success})
    for gtol in [1e-2, 1e-5, 1e-8, 1e-12]:
        opt = model.solve(do_print=False, options={'gtol': gtol})
        rows.append({'Calibration': name, 'Setting': 'gtol', 'Value': gtol,
                      's1': opt.s1, 'w': opt.w, 'u': opt.u,
                      'Evaluations': opt.res.nfev, 'success': opt.res.success})

df_tol = pd.DataFrame(rows)
df_tol
```

**Which settings would we use, and why is a tighter tolerance not always a better
answer?**

The table shows that both `ftol` and `gtol` give essentially identical solutions and utility
once they drop below about $10^{-8}$ (ftol) or $10^{-5}$-$10^{-8}$ (gtol) -- the last, tighter
steps ($10^{-12}$, $10^{-16}$) cost extra function evaluations without changing the solution
noticeably, because the solver is already at the limit of what the numerical gradient
calculation can resolve. A tighter tolerance is therefore not automatically "more correct":
once the tolerance drops below the numerical noise in the finite-difference gradient (step size
$\approx 10^{-8}$), we're pushing the solver to "converge" on noise rather than on real
improvements in $u$ -- we pay for iterations that don't actually move the answer. Since we are
maximizing, we can rank the settings without knowing the true answer: pick the loosest
tolerance that still gives the highest achieved $u$ -- here `ftol=1e-8` and `gtol=1e-5` are a
sensible choice: they give the same utility as the tightest settings, at a fraction of the
cost.

## Summary

* Grid search and L-BFGS-B find the same solution in both calibrations (check 2 from section
  1.3), but L-BFGS-B does it with roughly $10{,}000$ times fewer function evaluations.
* The grid solution converges smoothly in $u$, but unevenly in $(s_1,w)$, as $N$ increases --
  clear diminishing returns from $N=500$ to $N=1000$.
* The complements calibration ($\sigma_B=0.40$) has a steeper, more funnel-shaped utility
  surface than the substitutes calibration ($\sigma_B=3.00$), and it is also in the
  complements calibration that two of the corner starting points trick L-BFGS-B into a false
  `success = True` -- a concrete example of why checks 1 and 2 from section 1.3 matter, not
  just as formalities.
* Looser tolerances (`ftol=1e-8`, `gtol=1e-5`) give the same answer as much tighter
  tolerances, but more cheaply.
* The convergence path is slow and roughly linear while far from the optimum, and accelerates sharply (more than an order of magnitude per iteration) once L-BFGS-B is close to it -- the textbook superlinear-*near-the-solution* behaviour of a quasi-Newton method, not the reverse.

# Part 3: Relative prices and demand

# 3.1

**Question:** solve the model for $p_3$ from 0.5 to 3.0, holding $p_1$, $p_2$ and $I$ fixed, and plot the three budget shares and the three quantities, for both calibrations.

**Approach:** for each calibration we build a grid of 100 train prices, re-solve the consumer's problem with L-BFGS-B at every price (everything else held fixed), and read off the resulting shares $(s_1,s_2,s_3)$ and quantities $(x_1,x_2,x_3)$. This is a straightforward loop over `model.solve()` -- nothing new is needed, since `Consumer.py` already lets us change `model.par.p3` and re-solve.

```python
import numpy as np
import matplotlib.pyplot as plt

from Consumer import ConsumerClass

# two consumers, one per calibration -- only sigma_B differs, exactly as in the table
model_comp = ConsumerClass()
model_comp.par.sigma_B = 0.4  # complements

model_sub = ConsumerClass()
model_sub.par.sigma_B = 3.0   # substitutes

# store the two models in a dictionary for easy access
models = {
    "Complements": model_comp,
    "Substitutes": model_sub
}

# grid of train prices to sweep over, holding p1, p2 and I fixed
p3_grid = np.linspace(0.5, 3.0, 100)

# for each calibration and each p3, re-solve the consumer's problem with
# L-BFGS-B and store the resulting shares and quantities
results = {}

for name, model in models.items():

    s1 = []
    s2 = []
    s3 = []

    x1 = []
    x2 = []
    x3 = []

    I = model.par.I
    p1 = model.par.p1
    p2 = model.par.p2

    for p3 in p3_grid:

        model.par.p3 = p3          # change only the train price ...
        opt = model.solve(do_print=False)  # ... and re-solve from scratch

        s1.append(opt.s1)
        s2.append(opt.s2)
        s3.append(opt.s3)

        # quantities follow directly from the shares: x_j = s_j*I/p_j
        x1.append(opt.s1 * I / p1)
        x2.append(opt.s2 * I / p2)
        x3.append(opt.s3 * I / p3)

    results[name] = {
        "s1": np.array(s1),
        "s2": np.array(s2),
        "s3": np.array(s3),
        "x1": np.array(x1),
        "x2": np.array(x2),
        "x3": np.array(x3)
    }

# --- Figure 1: the three budget shares against the train price ---
fig, axes = plt.subplots(1, 2, figsize=(12,5), sharey=True)

for ax, (name, res) in zip(axes, results.items()):

    ax.plot(p3_grid, res["s1"], label="$s_1$ (food)")
    ax.plot(p3_grid, res["s2"], label="$s_2$ (bus)")
    ax.plot(p3_grid, res["s3"], label="$s_3$ (train)")

    ax.set_title(name)
    ax.set_xlabel("$p_3$ (train price)")
    ax.set_ylabel("Budget share")
    ax.grid(alpha=0.3)
    ax.legend()

plt.tight_layout()
plt.show()

# --- Figure 2: the three quantities against the train price ---
fig, axes = plt.subplots(1, 2, figsize=(12,5), sharey=True)

for ax, (name, res) in zip(axes, results.items()):

    ax.plot(p3_grid, res["x1"], label="$x_1$ (food)")
    ax.plot(p3_grid, res["x2"], label="$x_2$ (bus)")
    ax.plot(p3_grid, res["x3"], label="$x_3$ (train)")

    ax.set_title(name)
    ax.set_xlabel("$p_3$ (train price)")
    ax.set_ylabel("Quantity")
    ax.grid(alpha=0.3)
    ax.legend()

plt.tight_layout()
plt.show()
```

# 3.2
**Complements** $(σ_B=0.4)$: A higher train price raises the train's budget share. Although consumers buy fewer train tickets, the price increase more than offsets the reduction in quantity, causing expenditure on trains to increase.

**Substitutes** $(σ_B=3)$: A higher train price lowers the train's budget share. Consumers substitute away from trains toward buses, so the reduction in train demand is large enough to outweigh the price increase, reducing expenditure on trains.

This is read directly off the *left-hand* figure above (budget shares): the $s_3$ line slopes up in the complements panel and down in the substitutes panel.

**Why:** within a CES nest, the elasticity of substitution $\sigma_B$ is exactly what governs this. When $\sigma_B<1$, bus and train are hard to substitute for each other, so a given percentage rise in $p_3$ triggers a *smaller* percentage fall in $x_3$ — expenditure $p_3x_3$, and so the budget share $s_3$, rises. When $\sigma_B>1$, substitution is easy, the percentage fall in $x_3$ is *larger* than the percentage rise in $p_3$, and $s_3$ falls. $\sigma_B=1$ (Cobb-Douglas) would be the knife-edge case where the budget share stays exactly constant as $p_3$ changes.

# 3.3

**Question:** what happens to the number of bus trips when train tickets get more expensive, and how do the two calibrations differ?

This is read off the $x_2$ (bus) line in the *quantities* figure from 3.1 -- we do not need to solve anything new, only look at the same results from a different angle.

**Complements** $(σ_B=0.4)$: As train tickets become more expensive, the number of bus trips decreases. Since buses and trains are complementary goods, a reduction in train travel also reduces the demand for bus travel.

**Substitutes** $(σ_B=3)$: As train tickets become more expensive, the number of bus trips increases. Consumers substitute away from trains and towards buses, leading to higher bus demand.

# 3.4

**Question:** how much does the price of a train ticket change the demand for food, compared with its effect on bus trips?

Again we read this off the quantities figure from 3.1, this time comparing the $x_1$ (food) line with the $x_2$ (bus) line.

The price of a train ticket has only a small effect on food demand in both calibrations. Food consumption decreases slightly as train prices increase, reflecting a modest income effect rather than substitution. In contrast, the effect on bus trips is much larger. With complements $(σ_B=0.4)$, bus demand decreases together with train demand, whereas with substitutes $(σ_B=3)$, bus demand increases substantially as consumers substitute buses for trains.

This is exactly what the nested structure predicts: $\sigma_A$ (food vs. the whole travel composite) governs how much of the adjustment to a train-price change spills over into food, while $\sigma_B$ (bus vs. train) governs how it is split *within* the travel nest. Since $\sigma_A=0.80$ is the same, and close to 1, in both calibrations, a price change confined to the travel nest has only a modest knock-on effect on food in either case — most of the adjustment happens inside the travel nest itself, between bus and train, exactly where $\sigma_B$ differs between the two calibrations.

## 4. Lump-sum taxes and product taxes

The government can use a lump-sum tax or taxes on food, bus and train.  
For product taxes, the tax changes the price paid by the consumer. The consumer then chooses again.

The main idea is to compare taxes with the **same revenue**. A tax is better if the consumer keeps more utility.

```python
import numpy as np
import matplotlib.pyplot as plt

# GovernmentClass inherits from ConsumerClass (Government.py), so every method
# from Consumer.py -- .solve(), .shares(), .quantities(), .value_of_choice() --
# keeps working unchanged once taxes are set with .set_taxes()
from Consumer import ConsumerClass
from Government import GovernmentClass
```

### 4.1 Tax revenue

`tax_revenue()` is implemented in `Government.py` from equation (5):

\[
R=T+\tau_1p_1x_1+\tau_2p_2x_2+\tau_3p_3x_3.
\]

The quantities are the quantities the consumer chooses **after the tax**.

```python
# quick manual check of tax_revenue(): tax only the bus at 20%, solve, and
# read off the revenue -- it should equal tau2 * p2_pre * x2 (bus tax rate
# times the pre-tax bus price times the quantity of bus trips actually bought)
gov = GovernmentClass()

gov.set_taxes(tau2=0.20)          # raises p2 to 1.20 * p2_pre, everything else unchanged
opt = gov.solve(do_print=False)   # the consumer re-optimizes at the new (higher) bus price

print("Tax revenue:", gov.tax_revenue(opt))
```

### 4.2 Revenue as a function of the tax rate

I use the five product taxes from the question:

1. food only,
2. bus only,
3. train only,
4. bus and train,
5. all three goods.

I do this for the baseline calibration and the substitutes calibration.  
The tax rate goes from 0 to 3 as required in the question.

```python
# the five product-tax combinations asked for in the question: which good(s)
# get the same tax rate tau (goods are numbered 1=food, 2=bus, 3=train)
taxes = {
    "Food": (1,),
    "Bus": (2,),
    "Train": (3,),
    "Bus + train": (2,3),
    "All goods": (1,2,3)
}

# the two calibrations from the table in section 1.2 -- alpha, beta and the
# prices/income stay at their defaults, only sigma_A and sigma_B are set here
calibrations = {
    "Complements": {"sigma_A": 0.80, "sigma_B": 0.40},
    "Substitutes": {"sigma_A": 0.80, "sigma_B": 3.00}
}

tau_grid = np.linspace(0,3,101)  # tax rates from 0 to 3, as required by the question

# revenue_results[calibration][tax] holds the revenue curve (one value per tau in tau_grid)
revenue_results = {}

for name,par in calibrations.items():

    gov = GovernmentClass(par)
    revenue_results[name] = {}

    for tax_name,goods in taxes.items():

        revenue = []

        for tau in tau_grid:
            # revenue_and_utility() sets the tax, re-solves the consumer's
            # problem, and returns both the revenue and the utility reached --
            # here we only need the revenue
            R,u = gov.revenue_and_utility(tau,goods)
            revenue.append(R)

        revenue_results[name][tax_name] = np.array(revenue)
```

```python
# one figure, one panel per calibration, side by side -- each panel scaled
# to its own data (squished per-panel y-limits) so a low peak (e.g. the bus
# or train tax under substitutes) is still clearly visible, instead of being
# flattened by a y-axis sized for the tallest curve in the *other* panel
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5), sharex=True)

for ax, name in zip(axes, calibrations):

    for tax_name in taxes:
        ax.plot(tau_grid, revenue_results[name][tax_name], label=tax_name)

    panel_ymax = max(revenue_results[name][t].max() for t in taxes)

    ax.set_xlim(0, 3)
    ax.set_ylim(0, panel_ymax * 1.05)
    ax.set_xlabel("Tax rate")
    ax.set_ylabel("Revenue")
    ax.set_title(name)
    ax.legend(fontsize=8)

plt.tight_layout()
plt.show()
```

The curves show that product taxes do not all behave in the same way.
For some taxes, revenue gets a clear top because demand falls when the tax becomes high. Other curves keep increasing in the interval from 0 to 3.

The substitutes calibration reacts more to relative prices. This is especially important for the tax on train tickets because consumers can move towards the bus when train becomes expensive.

**Why:** a tax on a good with an easy substitute available (the train, when $\sigma_B=3$) erodes its own base quickly as the rate rises — consumers simply switch to the bus — so the Laffer-type peak in revenue is reached at a comparatively low rate. A tax with (almost) nothing for demand to shift to, like the lump-sum tax or a broad tax on all three goods at once, keeps raising revenue over a much wider range, because there is no substitution margin left to erode it — only the income effect can eventually choke off demand.

### 4.3 Maximum revenue

The hint says to use a grid over the tax rate and pick the best point.  
This is what `max_revenue()` does.

I use a larger grid than in the first figure because a maximum can be outside the interval 0 to 3.

```python
# for each instrument, find the revenue-maximizing tax rate with a grid
# search over tau (max_revenue() in Government.py), using a wide grid
# (0 to 10) since some curves in the figure above were still rising at tau=3
for name,par in calibrations.items():

    print("\n",name)

    gov = GovernmentClass(par)

    for tax_name,goods in taxes.items():

        tau_max,R_max = gov.max_revenue(
            goods=goods,
            tau_max=10,
            N=1001
        )

        # if the best point found is right at the edge of the grid (tau=10),
        # the curve had not yet turned over -- there is no top in this range
        if np.isclose(tau_max,10):
            print(tax_name, ": still rising at 10")
        else:
            print(
                tax_name,
                ": max tax =", round(tau_max,3),
                ", max revenue =", round(R_max,3)
            )
```

A top means that a higher tax eventually reduces demand so much that revenue starts to fall.

The train tax is the interesting case. When bus and train are stronger substitutes, a high train tax makes it easier for the consumer to move to bus. This can limit how much revenue the train tax can raise.

### 4.4 Utility against revenue

Now I compare the instruments by the utility left for the consumer at each level of revenue.  
I also include the lump-sum tax.

The x-axis is revenue, so taxes that are higher in the figure are better at raising the same revenue.

```python
# for each instrument (the five product taxes, plus the lump-sum tax), trace
# out the (revenue, utility) pairs as the tax rate rises from 0 to 3 -- this
# is the data behind the "utility against revenue" figure below
utility_results = {}

for name,par in calibrations.items():

    gov = GovernmentClass(par)
    utility_results[name] = {}

    # product taxes: same loop structure as the revenue figure above, but now
    # keeping both R and u from revenue_and_utility()
    for tax_name,goods in taxes.items():

        R_list = []
        u_list = []

        for tau in tau_grid:
            R,u = gov.revenue_and_utility(tau,goods)
            R_list.append(R)
            u_list.append(u)

        utility_results[name][tax_name] = (
            np.array(R_list),
            np.array(u_list)
        )

    # lump-sum tax: same idea, but sweeping T instead of tau, using
    # revenue_and_utility_lump_sum()
    T_grid = np.linspace(0,3,101)
    R_list = []
    u_list = []

    for T in T_grid:
        R,u = gov.revenue_and_utility_lump_sum(T)
        R_list.append(R)
        u_list.append(u)

    utility_results[name]["Lump sum"] = (
        np.array(R_list),
        np.array(u_list)
    )
```

```python
# one figure, one panel per calibration, side by side. Within a panel, a
# higher curve at a given revenue (x-axis) means the consumer keeps more
# utility there -- i.e. that instrument is the cheaper way to raise that
# revenue. The two panels share the same axes, so the *baseline* utility
# level (u at revenue=0) can also be compared directly across calibrations.
all_R = [R for name in utility_results for R, u in utility_results[name].values()]
all_u = [u for name in utility_results for R, u in utility_results[name].values()]
xmin, xmax = min(r.min() for r in all_R), max(r.max() for r in all_R)
ymin, ymax = min(u.min() for u in all_u), max(u.max() for u in all_u)

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5), sharey=True)

for ax, name in zip(axes, calibrations):

    for instrument, (R, u) in utility_results[name].items():
        ax.plot(R, u, label=instrument)

    ax.set_xlim(xmin, xmax)
    ax.set_ylim(ymin, ymax)
    ax.set_xlabel("Revenue")
    ax.set_ylabel("Utility")
    ax.set_title(name)
    ax.legend(fontsize=8)

plt.tight_layout()
plt.show()
```

The lump-sum tax is normally the cheapest way to raise revenue because it reduces income but does not change relative prices.

Product taxes also change what the consumer buys. This creates an extra loss in utility.
The ranking between the product taxes can change with the substitution parameters because the consumer reacts differently to price changes.

**Why:** a product tax does two things at once — it takes money away from the consumer, exactly like the lump-sum tax, but it also distorts the relative prices they face, pushing their bundle away from what they would otherwise choose at that income level. That second effect is a pure efficiency loss with no offsetting revenue for the government — the deadweight loss familiar from public economics. The lump-sum tax has only the first effect, so for a given amount of revenue it always leaves the consumer at least as well off as any product tax does. How large the gap is depends on $\sigma_B$: the easier bus and train are to substitute, the more a train-only tax distorts the bundle away from the consumer's free choice, and the bigger the utility gap to the lump-sum tax becomes.

**Why can some instruments only raise a limited amount of revenue?** In the substitutes calibration, a bus-only tax peaks at $R\approx0.87$ (at $\tau\approx0.86$) and a train-only tax peaks at just $R\approx0.26$ (at $\tau\approx0.59$) — both *fall* after that as $\tau$ keeps rising (section 4.3). Every other instrument, in both calibrations, is still climbing at $\tau=10$. The reason is the same substitution margin discussed above: when its close substitute is left untaxed, the quantity of the taxed good is driven toward zero as $\tau\to\infty$, and revenue $=\tau\cdot p\cdot x$ is a rate that keeps growing times a quantity that shrinks even faster — so revenue eventually falls back toward zero. A tax with no available substitute inside the model (food, or all three goods at once, or the lump-sum tax) has no such margin: the taxed quantity shrinks only slowly, so a rising $\tau$ keeps winning and revenue keeps climbing over the whole range shown.

**That is also the "odd shape":** the bus and train curves in the substitutes panel are the only two with a genuine interior peak (a Laffer curve) — they rise, top out, and then fall — while every other curve in both panels is monotonically increasing over $[0,3]$. A single hump instead of a steady climb looks "odd" next to the other four curves in the same figure, but it is exactly what a Laffer curve looks like once a close, untaxed substitute is available.

**Why doesn't utility ever reach exactly 0?** Utility only goes to zero in the limit of an infinite tax, not at any finite rate: at $\tau=10$ on "all goods" utility is already down to $0.31$ (from a baseline of $3.40$), at $\tau=100$ it is $0.03$, and it keeps shrinking without ever quite hitting zero, because the CES aggregator stays strictly positive for any strictly positive (even tiny) quantity. Revenue behaves the same way in reverse — it approaches the consumer's whole income $I=10$ as $\tau\to\infty$, but never reaches it exactly, for the same reason.

### 4.5 Raise exactly R = 0.20

The question asks for a root-finder.  
`find_tax_rate()` uses `optimize.root_scalar(..., method='brentq')`, as suggested in the hint.

For the lump-sum tax the answer is simple: `T = 0.20`.

For product taxes I first look for the low tax rate that raises 0.20. If an instrument cannot raise the target, the result is `nan`.

```python
R_target = 0.20

for name,par in calibrations.items():

    print("\n",name)

    gov = GovernmentClass(par)

    # lump-sum tax: T directly equals the revenue target, no root-finding needed
    R_lump,u_lump = gov.revenue_and_utility_lump_sum(R_target)

    print(
        "Lump sum:",
        "tax =", round(R_target,4),
        "utility =", round(u_lump,4)
    )

    # product taxes: find_tax_rate() brackets the root in (1e-10, 3.0) and
    # uses Brent's method (optimize.root_scalar) to find the rate that raises
    # exactly R_target -- if no such rate exists in the bracket (the target is
    # above what the instrument can ever raise, cf. section 4.3), it returns nan
    for tax_name,goods in taxes.items():

        tau = gov.find_tax_rate(
            R_target,
            goods=goods,
            bracket=(1e-10,3.0)
        )

        if np.isnan(tau):
            print(tax_name, ": cannot raise 0.20 in this bracket")

        else:
            # re-solve once more at the found rate, just to report the utility
            R,u = gov.revenue_and_utility(tau,goods)

            print(
                tax_name,
                ": tax =", round(tau,4),
                "utility =", round(u,4)
            )
```

```python
# Train, Substitutes comes back "cannot raise 0.20 in this bracket" above --
# but its peak revenue (section 4.3) is about 0.26, which is *above* 0.20. So
# the target should be reachable. What is actually going on?
gov = GovernmentClass({'sigma_A': 0.80, 'sigma_B': 3.00})

R0, _ = gov.revenue_and_utility(1e-10, (3,))
R3, _ = gov.revenue_and_utility(3.0, (3,))
print(f"revenue at the two ends of the bracket used by find_tax_rate: "
      f"R(1e-10)={R0:.4f}, R(3.0)={R3:.4f} -- both below the target 0.20, "
      f"so there is no sign change over the *whole* bracket and brentq cannot start.")

# but a target of 0.20 IS below the peak of ~0.26 (section 4.3), so there
# must be two rates that hit it: one on the way up, one on the way down.
# find each one by bracketing around the peak instead of the whole range.
tau_peak, R_peak = gov.max_revenue(goods=(3,), tau_max=3.0, N=2001)
tau_low = gov.find_tax_rate(0.20, goods=(3,), bracket=(1e-10, tau_peak))
tau_high = gov.find_tax_rate(0.20, goods=(3,), bracket=(tau_peak, 3.0))
print(f"peak revenue {R_peak:.4f} at tau={tau_peak:.3f}")
print(f"the two rates that actually raise R=0.20: tau={tau_low:.3f} (rising side), "
      f"tau={tau_high:.3f} (falling side)")
```

The table makes the comparison direct because every instrument raises the same revenue.

The instrument with the highest utility is the best one in this comparison.
The train tax is especially interesting because its result can change a lot when bus and train become stronger substitutes. If people can easily move from train to bus, the train tax becomes a weaker way to raise revenue.

**"Cannot raise 0.20" is not the same as "cannot raise 0.20 at all".** Train tax, substitutes calibration, is reported above as unreachable, but its peak revenue is about 0.26 (section 4.3) — above the target. The code cell just above shows why: revenue is *below* 0.20 at both ends of the bracket `find_tax_rate` searches, $(10^{-10},3.0)$, because the curve rises above 0.20 and falls back below it again in between (its Laffer peak, section 4.4). `optimize.root_scalar(..., method='brentq')` needs the function to have opposite signs at the two ends of the bracket it is given, and here it does not — even though two solutions genuinely exist inside that bracket. Narrowing the bracket to either side of the peak recovers both of them: a low rate on the rising part of the curve, and a much higher one on the falling part, where the tax has already driven quantity down a lot and needs an even higher rate to recover the same revenue on a shrunken base. This is the same lesson as section 2.2.3 in reverse: the bracket (there, the starting point) decides which answer — or whether any answer — a numerical method reports, and a `nan`/`ValueError` has to be read in context rather than taken as the final word.

**Are the utility costs different between the two calibrations?** Yes, and by more than the utility levels alone suggest. The *spread* across instruments — the gap between the cheapest (lump-sum) and the most expensive product tax at this revenue target — is roughly $0.001$-$0.0015$ in the complements calibration (all instruments cost almost exactly the same), but about $0.003$ in the substitutes calibration, with the bus tax the clear outlier. That matches section 4.4's deadweight-loss argument directly: at a small revenue target like $R=0.20$, the distortion from any one instrument is a second-order effect, so *which* good gets taxed barely matters when substitution is hard (complements). It matters much more when substitution is easy (substitutes), because that is exactly the channel — the ability to shift consumption away from the taxed good — that turns a tax into an efficiency loss on top of the revenue it raises.

# 5. Extension – a revenue-neutral green transport reform

For my extension I add a **green transport reform**.

The government puts a tax on bus trips, but instead of keeping the revenue, it uses all of it to make train tickets cheaper. This is different from task 4, where the tax revenue leaves the model.

I assume:
- one bus trip has an emissions index of **1.0**,
- one train trip has an emissions index of **0.30**.

For every bus tax, I use a root-finder to find the train subsidy that makes total product-tax revenue equal to zero.

This gives a simple question: **Can the government move travel from bus to train without using extra money?**

### 5.1 Baseline before the reform

```python
# the situation with no taxes at all, for both calibrations -- our reference
# point for "before the reform"
green_calibrations = {
    "Complements": {"sigma_B": 0.40},
    "Substitutes": {"sigma_B": 3.00},
}

baseline = {}
for name, par in green_calibrations.items():

    gov = GovernmentClass(par)
    gov.set_taxes()                          # all taxes/subsidies at zero
    base = gov.solve(do_print=False)
    x1_b, x2_b, x3_b = gov.quantities(base.s1, base.w)

    # a simple linear emissions index: one bus trip pollutes more than one
    # train trip (1.0 vs 0.30), so total emissions are a weighted sum of the two
    emissions_b = 1.0 * x2_b + 0.30 * x3_b

    baseline[name] = {
        'gov': gov, 'u': base.u, 'bus': x2_b, 'train': x3_b, 'emissions': emissions_b,
    }

    print(f"{name}: utility={base.u:.4f}, bus={x2_b:.4f}, train={x3_b:.4f}, "
          f"emissions={emissions_b:.4f}")
```

### 5.2 Try different bus taxes

For each bus tax, `green_reform()` finds the train subsidy that balances the government budget.

```python
# sweep the bus tax rate from 0 to 1 for both calibrations; for each rate,
# green_reform() (in Government.py) uses a root-finder to pick the train
# subsidy that makes the government's total product-tax revenue exactly zero
# -- i.e. the bus tax fully (and only) funds the train subsidy
bus_tax_grid = np.linspace(0, 1.0, 51)

green_results = {}
for name in green_calibrations:

    gov = baseline[name]['gov']

    subsidy, utility, bus, train, emissions = [], [], [], [], []
    for tau_bus in bus_tax_grid:
        result = gov.green_reform(tau_bus)
        subsidy.append(result.subsidy_train)
        utility.append(result.u)
        bus.append(result.x2)
        train.append(result.x3)
        emissions.append(result.emissions)

    green_results[name] = {
        'subsidy': np.array(subsidy), 'utility': np.array(utility),
        'bus': np.array(bus), 'train': np.array(train), 'emissions': np.array(emissions),
    }
```

```python
# Figure: how large a train subsidy does a given bus tax buy, in each calibration?
fig, ax = plt.subplots()
for name in green_calibrations:
    ax.plot(bus_tax_grid, green_results[name]['subsidy'], label=name)
ax.set_xlim(0, 1)
ax.set_xlabel("Tax on bus")
ax.set_ylabel("Subsidy on train")
ax.set_title("Balanced green reform")
ax.legend()
plt.show()
```

The required train subsidy increases with the bus tax in both calibrations, but by very different amounts. At $\tau_{bus}=1.0$ the subsidy needed to balance the budget is about **46%** off the pre-tax train price in the complements calibration, but only about **16%** in the substitutes calibration.

**In terms of the actual prices the consumer faces:** with $\tau_{bus}=1.0$, the bus price always doubles from $p_2=1.0$ to $p_2=2.0$ (that part is mechanical and the same in both calibrations). What differs is the train price: it falls from $p_3=1.5$ to about **0.80** in the complements calibration, but only to about **1.25** in the substitutes calibration. The reason is the tax base the reform has to work with: in the substitutes baseline the consumer already buys a lot of bus (3.20) and comparatively little train (0.95), so a 100% bus tax raises a large amount of revenue relative to how much a train subsidy of a given size costs — a smaller subsidy rate is enough to spend it all. In the complements baseline bus and train are bought in more similar quantities (2.04 vs 1.74), and the consumer does not abandon bus as readily when it gets taxed, so a bigger subsidy rate is needed to spend the same tax revenue on a train quantity that is itself changing as the subsidy grows.

```python
# Figure: how does the reform shift travel demand between bus and train,
# in each calibration?
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5), sharex=True, sharey=True)
for ax, name in zip(axes, green_calibrations):
    ax.plot(bus_tax_grid, green_results[name]['bus'], label="Bus")
    ax.plot(bus_tax_grid, green_results[name]['train'], label="Train")
    ax.set_xlim(0, 1)
    ax.set_xlabel("Tax on bus")
    ax.set_ylabel("Trips")
    ax.set_title(name)
    ax.legend()
plt.tight_layout()
plt.show()
```

The reform shifts travel from bus to train in both calibrations, but the shift is much larger under substitutes: bus trips fall from 3.20 to 0.67 (a drop of 79%) and train trips rise from 0.95 to 2.71, against a much more modest shift under complements (bus 2.04→1.48, train 1.74→2.13). This is exactly the substitution-elasticity mechanism from sections 3-4: with $\sigma_B=3$, a given change in relative prices moves a lot of demand between bus and train; with $\sigma_B=0.40$, the consumer wants to keep buying both together and reacts much less.

```python
# Figure: net effect on the emissions index, compared with each calibration's
# own no-tax baseline (dashed line)
fig, ax = plt.subplots()
colors = plt.rcParams['axes.prop_cycle'].by_key()['color']
for i, name in enumerate(green_calibrations):
    ax.plot(bus_tax_grid, green_results[name]['emissions'], color=colors[i], label=name)
    ax.axhline(baseline[name]['emissions'], color=colors[i], linestyle="--",
               label=f"{name}, no reform")
ax.set_xlim(0, 1)
ax.set_xlabel("Tax on bus")
ax.set_ylabel("Emissions index")
ax.set_title("Emissions")
ax.legend(fontsize=8)
plt.show()
```

The emissions index falls in both calibrations as travel moves from bus to train, but the reform is far more effective under substitutes: at $\tau_{bus}=1.0$ emissions fall by about 2.00 (from 3.48 to 1.48) against only about 0.44 (from 2.56 to 2.12) under complements. A reform that works by shifting *quantities* between two goods is naturally more powerful exactly where those two goods are easy to shift between.

```python
# Figure: what does the reform cost the consumer in utility, compared with
# each calibration's own no-tax baseline (dashed line)?
fig, ax = plt.subplots()
for i, name in enumerate(green_calibrations):
    ax.plot(bus_tax_grid, green_results[name]['utility'], color=colors[i], label=name)
    ax.axhline(baseline[name]['u'], color=colors[i], linestyle="--",
               label=f"{name}, no reform")
ax.set_xlim(0, 1)
ax.set_xlabel("Tax on bus")
ax.set_ylabel("Utility")
ax.set_title("Consumer utility")
ax.legend(fontsize=8)
plt.show()
```

### 5.3 Conclusion

The green reform lowers the use of bus and increases the use of train in both calibrations, and because train has lower emissions in this simple index, total transport emissions fall. The reform is revenue-neutral by construction: all bus-tax revenue is used, and only used, for the train subsidy.

The comparison across calibrations is the main finding: the reform is much **more effective, but also more costly**, when bus and train are easy substitutes ($\sigma_B=3$). Under substitutes it moves almost 80% of bus trips to train and cuts emissions by more than half, but it also costs about three times as much utility (0.376 vs 0.132 at $\tau_{bus}=1.0$) as the same policy does under complements. That trade-off is exactly the deadweight-loss mechanism from section 4.4-4.5 showing up again: a bus tax distorts relative prices more, and does more real work, precisely where substitution is easy — the same property that makes the reform *work* is what makes it *expensive* to the consumer.

I think this is an interesting extension because it changes the role of the government compared with section 4, where the government only takes revenue out of the model. Here the government uses one tax to change the price of another good, and how well that works depends entirely on a parameter ($\sigma_B$) the government does not directly control.
